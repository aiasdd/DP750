### 🎯 데모: Azure Databricks 작업 영역 만들기

Azure Databricks 작업 영역은 여러 방법으로 만들 수 있습니다. Azure CLI와 PowerShell 방법을 살펴보겠습니다.

**Azure CLI 사용:**

```bash
# Create a resource group (if needed)
az group create --name myresourcegroup --location westus2

# Create an Azure Databricks workspace
az databricks workspace create \\
    --resource-group myresourcegroup \\
    --name mydatabricksws \\
    --location westus2 \\
    --sku premium
```

**PowerShell 사용:**

```powershell
# Create an Azure Databricks workspace
New-AzDatabricksWorkspace `
    -Name mydatabricksws `
    -ResourceGroupName myresourcegroup `
    -Location westus2 `
    -ManagedResourceGroupName databricks-managed-rg `
    -Sku premium
```

**핵심 매개 변수:**
- **요금 계층**: Standard(기본 기능), Premium(RBAC, Unity Catalog), 또는 Trial(14일 무료)
- **관리형 리소스 그룹**: Databricks 인프라 리소스를 위해 자동 생성되는 그룹

### 🎯 데모: SQL 워크로드

Databricks SQL을 사용하면 데이터 분석가가 익숙한 SQL 구문으로 데이터를 쿼리하고, 대시보드를 만들고, BI 분석을 수행할 수 있습니다.

**참고:** SQL 웨어하우스는 Premium 계층 이상에서 사용할 수 있습니다.

````python
%sql
-- SQL Analytics Example: Query Delta tables

-- Create a sample analysis query
SELECT 
    customer_region,
    YEAR(order_date) as order_year,
    COUNT(*) as total_orders,
    SUM(order_amount) as total_revenue,
    AVG(order_amount) as avg_order_value
FROM trainer_demo.demo_01.orders
WHERE order_date >= '2024-01-01'
GROUP BY customer_region, YEAR(order_date)
ORDER BY total_revenue DESC;

-- This query can be:
-- 1. Run in the SQL Editor and/or in this notebook
-- 2. Saved as a query for reuse
-- 3. Added to dashboards
-- 4. Scheduled to run automatically
````
