### ✅ 제공됨: 변환 모듈 생성

아래 셀을 실행하여 `transforms.py` 모듈을 `/tmp/lab12/`에 작성하세요. 이 모듈은 네 개의 함수를 제공합니다:

| 함수 | 설명 |
|---|---|
| `filter_valid_orders(df)` | `order_id` 또는 `customer_id`가 null인 행을 제거합니다 |
| `normalize_status(df)` | `status`를 소문자로 변환하고 공백을 제거합니다 |
| `calculate_total_with_tax(df, tax_rate=0.21)` | `total_with_tax` 컬럼을 추가합니다 |
| `run_order_pipeline(df, tax_rate=0.21)` | 세 가지 변환을 순서대로 실행합니다 |

코드를 주의 깊게 검토하세요 — 다음 연습에서 각 함수에 대한 테스트를 작성하게 됩니다.

```python
%%writefile /tmp/lab12/transforms.py
import pandas as pd


def filter_valid_orders(df):
    """order_id 또는 customer_id가 null인 주문을 제거합니다."""
    return df.dropna(subset=['order_id', 'customer_id'])


def normalize_status(df):
    """status 값을 소문자로 정규화하고 공백을 제거합니다."""
    df = df.copy()
    df['status'] = df['status'].str.lower().str.strip()
    return df


def calculate_total_with_tax(df, tax_rate=0.21):
    """주어진 세율을 기준으로 total_with_tax 컬럼을 추가합니다."""
    df = df.copy()
    df['total_with_tax'] = (df['total'] * (1 + tax_rate)).round(2)
    return df


def run_order_pipeline(df, tax_rate=0.21):
    """전체 주문 정리 파이프라인을 실행합니다."""
    df = filter_valid_orders(df)
    df = normalize_status(df)
    df = calculate_total_with_tax(df, tax_rate)
    return df
```
