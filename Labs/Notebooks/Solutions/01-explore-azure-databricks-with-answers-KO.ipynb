## 연습 2: SQL 명령 실행

Azure Databricks 노트북은 Python과 함께 SQL을 기본적으로 지원합니다. **매직 명령**(`%`로 시작하는 특수 지시문)을 사용하여 모든 셀의 언어를 전환할 수 있습니다. 예를 들어 `%sql`은 Databricks에 셀을 SQL로 해석하도록 지시합니다.

**시나리오:** CityMoves Transit 운영 팀은 주로 SQL로 작업합니다. Python을 작성하는 동일한 노트북에서 SQL이 원활하게 작동한다는 것을 보여주려고 합니다.

**과제:** 아래 셀에서 `%sql` 매직 명령을 사용하여 다음을 수행하는 SQL 문을 작성하세요:

1. 현재 날짜와 시간을 반환합니다.
2. `system_name`이라는 열로 `'CityMoves Transit'`과 같은 레이블을 반환합니다.

결과는 두 개 이상의 열이 있는 단일 행 테이블이어야 합니다.

> 💬 **Genie Code 팁:** Genie Code에 물어보세요:
> *"Databricks 노트북에서 현재 타임스탬프와 문자열 리터럴을 이름이 지정된 열로 반환하는 SQL SELECT 문을 작성해 줘"*
> 셀을 `%sql`로 시작하는 것을 잊지 마세요.

```python
%sql
SELECT
    'CityMoves Transit' AS system_name,
    current_timestamp() AS report_generated_at
```
